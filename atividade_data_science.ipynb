```python
# Exercício 19 - Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o ângulo entre eles (produto escalar normalizado)

# Nesse exercício eu gerei dois arrays com 20 números aleatórios entre -10 e 10
# e então usei algumas funções para fazer o cálculo do ângulo entre eles

# fiz o import
import numpy as np

# declarei as váriaveis
array1: list[ int ] = []
array2: list[ int ] = []

# gerei 20 números aletórios entre -10 e 10 com randint()
array1 = np.random.randint( -10, 10, 20 )
array2 = np.random.randint( -10, 10, 20 )

# fiz o cosseno usando dot() e linalg.norm() onde:
# dot() vai multiplicando os números ( primeiro pelo segundo, segundo pelo segundo, .... ) e no fim faz a soma dos resultados
# linalg.norm() pega o tamanho real do vetor
cosseno = np.dot( array1, array2 ) / ( np.linalg.norm( array1 ) * np.linalg.norm( array2 ) )

# depois calculei o ângulo usando arccos() e degress()
# arccos() devolve o ângulo em radiano
# degress() devolve em graus
# clip() eu usei só para devolver valor entre -1 e 1
angulo = np.degrees( np.arccos( np.clip( cosseno, -1, 1 ) ) )


# mostro os resultados com print() usando interpolação de string
print( f'Lista 1: { array1 }' )
print( f'Lista 2: { array2 }' )
print( f'O ângulo é: { angulo }' )
```

```python
# Exercício 01 - Calcule a integral da função f(x) = x² sin(x) no intervalo de 0 a π

# Nesse exercício eu usei o integrate dentro do SciPy para
# conseguir calcular a integral da função entre o intervalo
# solicitado, eu primeiro fiz uma função que faz o quadrado do x
# depois eu só usei o quad() passando 0 e pi para fazer o cálculo

# fim o import do integrate que é do scipy
from scipy import integrate

# fiz uma função para pegar o valor de x ao quadrado
def formula( x ):
  return x**2

# usei o quad() para fazer o cálculo para mim
# ele executa a minha função várias vezes substituindo o x pelos
# valores entre 0 e 3.14 e vai somando tudo em resultado com uma
# margem de erro ( por isso eu faço essa desestruturação )
resultado, erro = integrate.quad( formula, 0 , 3.14 )

# mostro o resultado com print() usando interpolação de string
print( f'O resultado da integral é: { resultado }' )
```

```python
# Exercício 02 - Resolva a equação diferencial dy/dx = x − y, com y(0) = 1, no intervalo de 0 a 10

#

# fiz o import do solve dentro de scipy.integrate
from scipy.integrate import solve_ivp

# fiz uma função para pegar o valor de x ao quadrado
def formula( x, y ):
  return x - y

# com o solve_ivp() eu passei a minha formula, o intervalo e o valor de y0 ( 1 )
# assim eu obtenho o resultado da equação diferencial
resultado = solve_ivp( formula, [ 0, 10 ], [ 1 ] )

# mostro o resultado com print() usando interpolação de string
print( f'O resultado da equação diferencial é: { resultado }' )
```

```python
# Exercício 03 - Encontre as raízes da função f(x) = x4 − 3x³ + 2 utilizando SciPy

# Nesse exercício comecei adpatando a minha formula para a do
# enunciado, depois eu fiz um array de chutes ( tentativas )
# onde ele vai procurar no "gráfico" se o x passa ou não do
# chão ( x = 0 ), por fim eu só usei o root() para ele fazer
# esse cálculo

# fiz o import do root dentro de scipy.optimize
from scipy.optimize import root

# fiz a formula
def formula( x ):
  return x ** 4 - 3 * x ** 3 + 2


# fiz uma lista de "chutes" para ele procurar
# em vários lugares do gráfico
chutes = [ -2, 0, 1, 3 ]

# executei o root() passando a formula e as tentativas
resultado = root( formula, chutes )

# mostro o resultado com print() usando interpolação de string
# coloquei resultado.x pois ele é onde ficam as respostas exatas
# dos chutes pois ele é o ponto onde o ponto cruza ou não o chão
print( f'O resultado das raízes da função são: { resultado.x }' )


```

```python
# Exercício 04 - Use a função de otimização de SciPy para maximizar f(x) = −x²+ 4x + 1

# Nesse exercício para maximizar a formula, eu tive que
# inverter a função ( onde ao invés de somar os valores eu
# subtraia ) e depois eu só usei o minimize() que na verdade
# maximiza a formula pois ela esta invertida. Fiz dessa forma
# pois o scipy não tem um metodo de maximize() por exemplo

# fiz o import do maximize do scipy.optimize
from scipy.optimize import minimize

# fiz a formula ao contrário para fazer o miminize() que
# na realidade vai maximizar pois a formula esta invertida
def formula( x ):
  # return ( -x ** 2 ) + ( 4 * x ) + 1
  return  ( x ** 2 ) - ( 4 * x ) - 1

# passei a formula e o defini o x como 0 para o método
# maximize()
resultado = minimize( formula, 0 )


# mostro o resultado com print() usando interpolação de string
print( f'O resultado do máximo encontrado em x é: { resultado.x }' )
```

```python
# Exercício 05 - Calcule a transformada de Fourier de um sinal composto de sin( 2πt ) e cos( 4πt )

# Nesse exercício usei tanto o NumPy como o SciPy para fazer o cálculo
# do seno e do coseno, depois de calcular eu só fiz a soma para ter o sinal
# composto e por fim transformei esse sinal usando o fft() do SciPy

# fiz o import do NumPy e do SciPy
import numpy as np
from scipy.fft import fft

# criei uma linha de números entre 0 e 1
t = np.linspace( 0, 1, 500 )

# usei o sin() e cos() para fazer o cálculo onde
# o seno é calculado multiplicando 2 vezes pelo pi e pelos números
# o coseno é calculado multiplicando 4 vezes pelo pi e pelos números
onda1 = np.sin( 2 * np.pi * t )
onda2 = np.cos( 4 * np.pi * t )

# depois eu fiz o sinal composto que é a onda1 mais a onda 2
sinal = onda1 + onda2

# e para transformar o resultado final eu só fiz o fft()
# onde ele aplica a Transformada de Fourier
resultado = fft( sinal )

# mostrei o resultado com print() usando interpolação de string
print( f'Resultado final: { resultado }' )

```

```python
# Exercício 06 - Encontre a matriz inversa de uma matriz 5x5 de números aleatórios

# Nesse exercício eu use o NumPy para fazer uma matriz 5x5 de
# números entre 1 e 10, depois eu usei o linalg() do SciPy para
# fazer a matriz inversa

# fiz o import do NumPy e do SciPy
import numpy as np
from scipy.linalg import inv

# fiz uma linha de 25 números entre 1 e 10
numeros = np.random.randint( 1, 10, 25 )

# depois usei o reshape para fazer a matriz 5x5
matriz = numeros.reshape( 5, 5 )

# depois inverti a matriz com inv()
inversa = inv( matriz )

# mostrei os resultados com print() usando interpolação de string
print( f'Matriz original: { matriz }' )
print( f'Matriz inversa: { inversa }' )
```

```python
# Exercício 07 - Use a função de interpolação cúbica de SciPy para interpolar os dados (0, 0), (1, 2), (2, 3), (3, 5), (4, 4)

# Nesse exercício eu comecei fazendo os arrays de X e de Y
# usando o NumPy e depois eu user o interp1d do SciPy para
# fazer a função de interpolação e assim poder retornar o
# resultado dos novos no fim

# fiz o import do NumPy e do SciPy
import numpy as np
from scipy.interpolate import interp1d

# fiz os arrays de X e de Y conforme os dados
x = np.array( [ 0, 1, 2, 3, 4 ] )
y = np.array( [ 0, 2, 3, 5, 4 ] )

# usei o interp1d para criar a função de interpolação
f = interp1d( x, y, kind = 'linear' )

# defini os novos X e Y
x_novos = np.array( [ 0.5, 1.5, 2.5 ] )
y_novos = f( x_novos )

# mostrei o resultado no final
print( f'Valores interpolados: { y_novos }' )
```

```python
# Exercício 08 - Resolva o sistema de equações 3x + 2y = 5 e x + 4y = 6

# Nesse exercício eu importei o NumPy e o SciPy e depois
# eu fiz as duas matrizes onde a primeira representa os valores
# de X e a segunda os valores de Y, depois eu usei o solve para
# resolver a equação

# fiz o import do NumPy e do SciPy
import numpy as np
from scipy.linalg import solve

# fiz as matrizes onde a primeira são os valores de X
# e a segunda são os valores de Y
matriz_a = np.array( [ [ 3, 2 ],
                       [ 1, 4 ] ] )
matriz_b = np.array( [ 5, 6 ] )

# usei o solve() para poder resolver a equação
resultado = solve( matriz_a, matriz_b )

# mostrei os resultados com print() usando interpolação de string
# a primeira posição é o valor de X e a segunda é o valor do Y
print( f'Resultado: { resultado }' )
print( f'X = { resultado[ 0 ] }' )
print( f'Y = { resultado[ 1 ] }' )
```

```python
# Exercício 09 - Ajuste uma curva quadrática aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8)

# Nesse exercício eu importei a NumPy e a SciPy e
# depois eu fiz os arrays de X e Y para poder passar
# para para a formula ( função que calcula a parabola )
# e por fim usei o curve_fit() para ajustar a curva com
# base nos dados

# fiz o import do NumPy e do SciPy
import numpy as np
from scipy.optimize import curve_fit

# fiz os arrays dos valores de X e Y
x = np.array( [ 1, 2, 3, 4, 5 ] )
y = np.array( [ 2, 4, 5, 7, 8 ] )

# fiz a função da parabola
def parabola( x, a, b, c ):
  return a * x ** 2 + b * x + c

# passei a função para o curve_fit() ajustar a curva
parametros, _ = curve_fit( parabola, x, y )

# mostrei o resultado com print() usando interpolação de string
print( f'Parâmetros: { parametros }' )

```

```python
# Exercício 10 - Calcule a integral dupla da função f(x, y) = xy + x² no intervalo [ 0, 2 ] para x e [ 0, 1 ] para y

# Nesse exercício eu importei o dblquad() para calcular a integral dupla
# depois eu fiz a formula e fiz as variaveis de inicio e fim de X e de Y
# depois eu passei esses valores para o dblquad()

# fim o import do integrate que é do scipy
from scipy.integrate import dblquad

# fiz a função com a formula
def formula( y, x ):
  return ( x * y ) + ( x ** 2 )

# fiz os X inicial e final
x_inicial = 0
x_fim = 2

# fiz os Y inicial e final
# o lambda serve para criar uma 'mini função'
y_inicial = lambda x: 0
y_fim = lambda x: 1

# calculei o resultado e a margem de erro usando o dblquad()
resultado, erro = dblquad( formula, x_inicial, x_fim, y_inicial, y_fim  )

# mostrei os resultados com print() usando interpolação de string
print( f'O resultado é: { resultado }' )
print( f'A margem de erro é de: { erro }' )
```

```python
# Exercício 12 - Resolva a equação ex − x² = 0 para encontrar a raiz mais próxima de x = 1

# Nesse exercício eu usei o NumPy para pegar o número de euler
# depois eu só usei o root() do SciPy para pegar a raiz mais
# próxima

# fiz o import do NumPy e do SciPy
import numpy as np
from scipy.optimize import root

# fiz a função passando a formula
def formula( x ):
  return np.exp( x ) - x ** 2

# procurei a raiz mais próxima usando root()
resultado = root( formula, 1 )

# mostrei o resultado com print() usando interpolação de string
print( f'Raiz encontrada: { resultado.x }' )
```

```python
# Exercício 14 - Utilize a função de otimização multivariada para minimizar f(x, y) = (x−2)² + (y − 3)²

# Nesse exercício eu usei o minimize() do SciPy
# para minimizar a função multivariada, na formula
# eu passei os valores com base no enunciado

# fiz o import do SciPy
from scipy.optimize import minimize

# fiz a função com a formula ( desestruturei o x e o y)
def formula( v ):
  x, y = v
  return ( x - 2 ) ** 2 + ( y - 3 ) ** 2

# fiz o cálculo usando a formula
resultado = minimize( formula, [ 0, 0 ] )

# mostrei o resultado com print() usando interpolação de string
print( f'Minimo em [ x, y ]: { resultado.x }' )
```

```python
# Exercício 15 - Calcule a transformada de Fourier inversa de um sinal complexo gerado aleatoriamente

# Nesse exercício eu usei o NumPy para gerar dois arrays
# de números aleatórios e fiz um sinal complexo usando um
# valor imaginário, depois eu só calculei o resultado inverso
# usando o ifft() do SciPy

# fiz o import do NumPy e do SciPy
import numpy as np
from scipy.fft import ifft

# fiz dois arrays de números aleatórios
array1 = np.random.rand( 10 )
array2 = np.random.rand( 10 )

# fiz o sinal usando os arrays que eu fiz
# o j representa uma parte imaginária ou seja
# é uma unidade imaginaria vezes o array2
sinal = array1 + ( 1j * array2 )

# fiz o resultado inverso usando ifft()
inverso = ifft( sinal )

# mostrei o resultado com print() usando interpolação de string
print( f'Sinal Complexo: { sinal }' )
print( f'Resultado inverso: { inverso }' )
```

```python
# Exercício 16 - Resolva a equação diferencial de segunda ordem d²y/dx² + y = 0, com condições iniciais de y(0) = 1 e y'(0) = 0

# Nesse exercício eu usei o solve_ivp() do SciPy para resolver a
# equação, primeiro eu fiz a formula onde eu "separei" os dois y
# e então eu só passei os valores para a função e fiz o cálculo

# fiz o import do SciPy
from scipy.integrate import solve_ivp

# fiz a função com a formula
# usei v ao invés de y pois é ele que tem os dois y
def formula( x, v ):
  y = v[ 0 ]
  y_linha = v[ 1 ]

  return [ y_linha, -y ]

# calculei usando o solve_ivp onde
# [ 0, 10 ] é o intervalo
# e o [ 1, 0 ] são as condições iniciais
resultado = solve_ivp( formula, [ 0, 10 ], [ 1, 0 ] )


# mostrei o resultado com print() usando interpolação de string
print( f'Valores de x: { resultado.t }' )
print( f'Resultados de y: { resultado.y[ 0 ] }' )

```

```python
# Exercício 18 - Crie um gráfico de barras com valores positivos e negativos e use diferentes cores para cada tipo de valor

# Nessse exercício fiz um array de valores positivos
# e negativos de alguns produtos, depois eu desenhei o
# gráfico fazendo uma condição que muda a cor da barra
# conforme ele for positivo ou negativo

# fiz os imports
import matplotlib.pyplot as plt
import numpy as np

# fiz alguns dados
produtos = [ 'Banco', 'Mesa', 'Cadeira', 'Prateleira' ]
valores = [ 1500, -300, 2000, -500 ]

# fiz uma condição nas cores onde caso seja positivo é verde
# e se for negativo é vermelho
cores = np.where( np.array( valores ) >=0, 'green', 'red' )

# desenhei o gráfico
plt.bar( produtos, valores, color = cores )
plt.title( 'Gráfico de barras' )
plt.xlabel( 'Produtos' )
plt.ylabel( 'Valores' )

plt.show()
```

```python
# Exercício 20 - Adicione uma grade com espaçamento personalizado a um gráfico de linha para y = x²

# Nesse exercício fiz um gráfico de linha, só
# que eu editei o espaçamento usando xticks()
# e o yticks() e editei a grade também

# fiz os imports
import matplotlib.pyplot as plt
import numpy as np

# fiz os valores de X e Y
x = np.linspace( -5, 5, 100 )
y = x ** 2

# fiz o gráfico e personalizei o espaçamento
# usando xticks() e yticks(), também editei o grid
plt.plot( x, y )

plt.xticks( np.arange( -5, 6, 1 ) )

plt.yticks( np.arange( 0, 30, 5 ) )

plt.grid( color = 'gray', linestyle = '--', linewidth = 0.5, alpha = 0.7 )

plt.title( 'Gráfico de linha' )
plt.xlabel('Eixo X')
plt.ylabel('Eixo Y')

plt.show()

```
